## Helper Functions

Event emitter for testing tool responses:

```python
# Event emitter to capture status messages during tool execution
events_log = []


async def test_event_emitter(event: Dict[str, Any]):
    """Captures events from tools for testing purposes"""
    events_log.append(event)
    event_type = event.get('type', 'unknown')
    data = event.get('data', {})

    if event_type == 'status':
        status = "✅ Done" if data.get('done') else "⏳ In Progress"
        print(f"[{event_type.upper()}] {status}: {data.get('description', 'No description')}")
    elif event_type == 'message':
        print(f"[MESSAGE] {data.get('content', 'No content')}")
    else:
        print(f"[{event_type.upper()}] {data}")


def clear_events_log():
    """Clear the events log"""
    global events_log
    events_log = []
    print("🧹 Events log cleared")


def print_result(result: str, title: str = "Result"):
    """Pretty print tool results"""
    print(f"\n{'=' * 50}")
    print(f"📋 {title}")
    print(f"{'=' * 50}")

    # Try to parse as JSON for better formatting
    try:
        parsed = json.loads(result)
        print(json.dumps(parsed, indent=2, ensure_ascii=False))
    except:
        print(result)

    print(f"{'=' * 50}\n")


print("✅ Helper functions ready")
```

## 7. Performance Test

Test the count parameter performance:

```python
import time

# Performance comparison: All vs Limited data
print("🚀 Performance Test: Comparing data retrieval times")
print("=" * 60)

# Test 1: Get all data
clear_events_log()
start_time = time.time()

result_all_perf = await tools.get_gto_data(
    gto_id=test_gto_id,
    count=-1,
    __event_emitter__=None  # Skip events for cleaner output
)

time_all = time.time() - start_time

# Test 2: Get limited data
start_time = time.time()

result_limited_perf = await tools.get_gto_data(
    gto_id=test_gto_id,
    count=5,
    __event_emitter__=None  # Skip events for cleaner output
)

time_limited = time.time() - start_time

# Compare results
try:
    all_data = json.loads(result_all_perf)
    limited_data = json.loads(result_limited_perf)

    all_count = len(all_data.get('data', [])) if all_data.get('success') else 0
    limited_count = len(limited_data.get('data', [])) if limited_data.get('success') else 0

    print(f"📊 All data: {all_count} entries in {time_all:.3f}s")
    print(f"📊 Limited data: {limited_count} entries in {time_limited:.3f}s")
    print(f"⚡ Speed improvement: {((time_all - time_limited) / time_all * 100):.1f}%" if time_all > 0 else "N/A")

except Exception as e:
    print(f"❌ Performance comparison failed: {e}")
    print(f"All data result: {result_all_perf[:200]}...")
    print(f"Limited data result: {result_limited_perf[:200]}...")

print("=" * 60)
```
