# Docling Ingestion Pipeline

```python
def inject_figures_into_md(md_file_path: str):
    """
    Load figures from a markdown file generated by docling.
    The figures are expected to be in base64 format within the markdown file.
    """
    with open(md_file_path, 'r') as file:
        md_content = file.read()

    # Extract base64 encoded images from markdown content
    import re
    figure_blocks = re.findall(r'!\[.*?\]\((.*?)\)', md_content)
    for figure_block in figure_blocks:
        figure_path = f"figures/{os.path.basename(figure_block)}"
        # load the image and convert it to base64
        with open(figure_path, 'rb') as file:
            image_data = file.read()
            base64_string = base64.b64encode(image_data).decode('utf-8')
            # replace the figure block with the base64 string
            md_content = md_content.replace(figure_block, f"data:image/png;base64,{base64_string}")

    with open(md_file_path, 'w') as file:
        file.write(md_content)
```

```python
import base64
import os

import requests

DOCLING_API_URL = "http://localhost:5001"
UPLOAD_ENDPOINT = f"{DOCLING_API_URL}/v1alpha/convert/source"

INPUT_FILE_PATH = "examples.pdf"
OUTPUT_FILE_PATH = "docling.md"


def encode_file_to_base64(file_path):
    """
    Encode a file to base64 string for API submission
    """
    try:
        with open(file_path, "rb") as file:
            encoded_string = base64.b64encode(file.read()).decode('utf-8')
        return encoded_string
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return None
    except Exception as e:
        print(f"Error encoding file: {str(e)}")
        return None


request_body = {
    "options": {
        "from_formats": [
            "docx",
            "pptx",
            "html",
            "image",
            "pdf",
            "asciidoc",
            "md",
            "csv",
            "xlsx",
            "xml_uspto",
            "xml_jats",
            "json_docling"
        ],
        "to_formats": [
            "md"
        ],
        "image_export_mode": "embedded",
        "do_ocr": True,
        "force_ocr": True,
        "ocr_engine": "easyocr",
        "pdf_backend": "dlparse_v4",
        "table_mode": "accurate",
        "abort_on_error": False,
        "return_as_file": False,
        "do_table_structure": True,
        "include_images": True,
        "images_scale": 2,
        "do_code_enrichment": True,
        "do_formula_enrichment": True,
        "do_picture_classification": False,
        "do_picture_description": False
    },
    "file_sources": [
        {
            "base64_string": encode_file_to_base64(INPUT_FILE_PATH),
            "filename": "test_file.pdf"
        }
    ]
}
```

```python
r = requests.post(
    UPLOAD_ENDPOINT,
    json=request_body,
    headers={"Content-Type": "application/json"},
    timeout=1000
)
answer = r.json()
```

```python
with open(OUTPUT_FILE_PATH, "a") as file:
    file.write(answer["document"]["md_content"])
```

```python

from llama_index.core.base.llms.types import ChatMessage, TextBlock, ImageBlock
import re

llm_config = AzureOpenAILLMConfig(
    name="gpt-4o",
    base_url="https://bbvaihub-openai-sui.openai.azure.com",
    api_version="2025-01-01-preview",
    prompt_tokens_costs_per_thousand=0.0045,
    completion_tokens_costs_per_thousand=0.0133,
)
llm, _ = llm_config.to_llama_index(llm_config.default_parameter)


def extract_blocks_from_md(md_file_path: str):
    with open(md_file_path, 'r') as file:
        md_content = file.read()

    text_blocks = re.split(r'!\[.*?\]\(.*?\)', md_content)
    base64_strings = re.findall(r'!\[.*?\]\((.*?)\)', md_content)
    text_blocks = [TextBlock(text=block.strip()) for block in text_blocks]
    image_bytes_list = []
    for base64_string in base64_strings:
        image_bytes_list.append(base64.b64decode(base64_string.split("data:image/png;base64,")[1]))
    figure_blocks = [ImageBlock(image=image_bytes) for image_bytes in image_bytes_list]
    text_blocks.extend(figure_blocks)
    return text_blocks


# di.md or docling.md
blocks = extract_blocks_from_md("docling.md")

for question in questions:
    temp_blocks = blocks.copy()
    temp_blocks.append(TextBlock(text=question))

    messages = [
        ChatMessage(
            role="system",
            content="Du bist ein hilfreicher Assistent, welcher fragen anhand von Dokumentinhalten beantwortet."
                    "Wenn du die Antwort nicht weisst, dann sag: 'Ich weiss es nicht.'. Formatiere die Antwort in Markdown."
        ),
        ChatMessage(
            role="user",
            blocks=temp_blocks
        )
    ]
    response = llm.chat(messages=messages)
    print(f"Frage: {question}\n\nAntwort: {response.message.content}\n\n\n\n")
```

Frage: Wie wird der Leistunsbedarf eines Ventilators berechnet?

Antwort: Der Leistungsbedarf eines Ventilators kann wie folgt berechnet werden:

\[ P_{\text{mains}} = \frac{q_v \cdot \Delta P_{\text{tot}}}{\eta_{\text{tot}}} \]

oder basierend auf dem statischen Druck:

\[ P_{\text{mains}} = \frac{q_v \cdot \Delta P_{\text{stat}}}{\eta_{\text{stat}}} \]

Dabei sind:
- \( P_{\text{mains}} \): Wirkleistung aus dem Stromversorgungsnetz, in Watt (W)
- \( q_v \): Luftvolumenstrom durch den Ventilator, in Kubikmeter pro Sekunde (m³/s)
- \( \Delta P_{\text{tot}} \): Gesamtdruckanstieg vom Ventilatoreintritt bis zum -austritt, in Pascal (Pa)
- \( \Delta P_{\text{stat}} \): Statischer Druckanstieg vom Ventilatoreintritt bis zum -austritt, in Pascal (Pa)
- \( \eta_{\text{tot}} \): Gesamtwirkungsgrad der Luftförderung, basierend auf dem Gesamtdruck
- \( \eta_{\text{stat}} \): Gesamtwirkungsgrad der Luftförderung, basierend auf dem statischen Druck

Der Gesamtwirkungsgrad (\(\eta_{\text{tot}}\) oder \(\eta_{\text{stat}}\)) setzt sich zusammen aus dem Wirkungsgrad des Ventilators, der mechanischen Kraftübertragung, des Elektromotors und der Steuer- und Regeleinrichtung.




Frage: Wie lautet die Formel für die Berechnung der Luftdichte?

Antwort: Die Dichte von trockener Luft kann über die Gasgleichung aus Druck und Temperatur berechnet werden:

\[
\rho = \frac{p}{R_s \cdot T}
\]

Für feuchte Luft kann die Dichte mit dem Feuchtegehalt \( x \) berechnet werden:

\[
\rho = \frac{p}{R_s \cdot T} \cdot \frac{0.622}{x + 0.622}
\]

Dabei ist:
- \( \rho \) die Luftdichte in kg/m³,
- \( p \) der Absolutdruck der Luft in Pa,
- \( R_s \) die spezifische Gaskonstante in J/(kg·K) (für trockene Luft ist \( R_s = 287.058 \) J/(kg·K)),
- \( T \) die absolute Lufttemperatur in K,
- \( x \) der Feuchtegehalt der Luft in kg/kg.

Antwort: Der Leistungsbedarf eines Ventilators kann mit der folgenden Formel berechnet werden:

\[ P_{\text{mains}} = \frac{q_v \cdot \Delta p_{\text{tot}}}{\eta_{\text{tot}}} = \frac{q_v \cdot \Delta p_{\text{stat}}}{\eta_{\text{stat}}} \]

Dabei sind:

- \( P_{\text{mains}} \): Wirkleistung aus dem Stromversorgungsnetz, in Watt (W)

- \( \Delta p_{\text{stat}} \): Statischer Druckanstieg vom Ventilatoreintritt bis zum -austritt, in Pascal (Pa)
- \( \eta_{\text{tot}} \): Gesamtwirkungsgrad der Luftförderung, basierend auf dem Gesamtdruck
- \( \eta_{\text{stat}} \): Gesamtwirkungsgrad der Luftförderung, basierend auf dem statischen Druck

Diese Formel berücksichtigt sowohl den Gesamtdruck als auch den statischen Druck, um den Leistungsbedarf des Ventilators zu bestimmen.

- \(R_s\) ist die spezifische Gaskonstante, für trockene Luft \(R_s = 287.058 \, \text{J/(kg-K)}\).
- \(T\) ist die absolute Lufttemperatur in K.
